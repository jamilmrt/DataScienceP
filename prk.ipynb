```python
import json
```

```python
def load_data(filename):
    with open(filename) as f:
        return json.load(f)
```

```python

def find_people(user_idd, dataa):
    user_friends = {}
    for _ in dataa['users']:
        user_friends  = {user['id']: set(user['friends'])
                         for user in dataa['users']}
    if user_idd not in user_friends:
        return []
    direct_friends = user_friends[user_idd]
    ssg ={}
    for friend in direct_friends:
        for mutual in user_friends.get(friend, []):
            if mutual != user_idd and mutual not in direct_friends:
                ssg[mutual] = ssg.get(mutual, [])
    sorted_ssg = sorted(ssg.items(), key= lambda x: x[1], reverse=True)
    # return [user_id for user_id in sorted_ssg]
    return [mutual_id for mutual_id, count in sorted_ssg]
```
