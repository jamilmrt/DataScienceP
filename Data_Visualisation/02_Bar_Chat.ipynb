```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Define 5 selected years and runs for each player
years = ['2008', '2009', '2010', '2011', '2012']
sachin = [1000, 850, 1500, 1100, 800]
kohli = [300, 500, 1000, 1300, 1000]
sehwag = [900, 1100, 1400, 1000, 700]

# 2. Set index positions and bar width
x = np.arange(len(years))  # [0, 1, 2, 3, 4]
width = 0.25

# 3. Create side-by-side bars
plt.figure(figsize=(10, 6))
plt.bar(x - width, sachin, width=width, label='Sachin', color='skyblue')
plt.bar(x, kohli, width=width, label='Kohli', color='orange')
plt.bar(x + width, sehwag, width=width, label='Sehwag', color='lightgreen')

# 4. Add labels, title, and legend
plt.xlabel('Years')
plt.ylabel('Runs Scored')
plt.title('Run Comparison Over 5 Selected Years')
plt.xticks(x, years)  # Replace index numbers with actual year labels
plt.legend()
plt.tight_layout()

# 5. Display plot
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Define players and calculate their total runs in debut 5 years
players = ['Sachin', 'Kohli', 'Sehwag']

# Example debut 5-year run totals
sachin_debut = [500, 700, 1100, 1500, 1800]
kohli_debut = [300, 500, 1000, 1300, 1000]
sehwag_debut = [200, 900, 1400, 1600, 1800]

total_runs = [sum(sachin_debut), sum(kohli_debut), sum(sehwag_debut)]

# 2. Create horizontal bar plot
plt.figure(figsize=(8, 5))
bars = plt.barh(players, total_runs, color=['skyblue', 'orange', 'lightgreen'])

# 3. Add values next to the bars using plt.text()
for bar in bars:
    width = bar.get_width()
    plt.text(
        width + 50,
        bar.get_y() + bar.get_height() / 2,
        f'{width}',
        va='center',
    )

# 4. Add labels and title
plt.xlabel('Total Runs Scored')
plt.ylabel('Players')
plt.title('Total Runs Scored in Debut 5 Years')
plt.tight_layout()

# 5. Display plot
plt.show()
```
