# Pie Charts in Matplotlib
- Pie charts are used to show part-to-whole relationships. They're visually appealing but best used with fewer categories, as too many slices can get cluttered and hard to interpret.

```python
import matplotlib.pyplot as plt

plt.style.use("ggplot")  # Choose any style you like
```

```python
# Data
labels = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs = [18000, 8000, 12000, 9500]
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']

plt.title("Career Runs of Indian Batsmen")
plt.pie(runs, labels=labels, colors=colors)
plt.show()
```

```python
plt.pie(
    runs,
    labels=labels,
    colors=colors,
    wedgeprops={'edgecolor': 'black', 'linewidth': 1, 'linestyle': '--'}
)
plt.show()
```

```python
explode = [0.1, 0.01, 0.01, 0]  # Only highlight Sachin's slice

plt.pie(runs, labels=labels, explode=explode, colors=colors)
plt.title("Exploded Pie Example")
plt.show()
```

```python
plt.pie(
    runs,
    labels=labels,
    explode=explode,
    colors=colors,
    shadow=True  # adds a 3D-like shadow
)
plt.show()
```

```python
plt.pie(
    runs,
    labels=labels,
    explode=explode,
    colors=colors,
    shadow=True,
    startangle=130
)
plt.show()
```

```python
plt.pie(
    runs,
    labels=labels,
    autopct='%1.2f%%',
    startangle=140
)
plt.title("Career Run Share")
plt.show()
```

```python
# Too many categories
languages = ['Python', 'Java', 'C++', 'JavaScript', 'C#', 'Ruby', 'Go', 'Rust', 'Swift', 'PHP']
usage = [30, 20, 10, 10, 7, 5, 4, 3, 2, 1]

plt.pie(usage, labels=languages, autopct='%1.1f%%', startangle=90)
plt.title("Programming Language Usage (Crowded Example)")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data: Mobile OS Market Share (%)
os_names = ['Android', 'iOS', 'Others']
market_share = [71.0, 28.0, 1.0]
colors = ['#3DDC84', '#555555', '#FF9999']  # Custom brand-like colors

# ==========================================
# 1. Pie Chart
# ==========================================
plt.figure(figsize=(6, 6))
plt.pie(
    market_share,
    labels=os_names,
    autopct='%1.1f%%',
    startangle=140,
    colors=colors,
    explode=[0.05, 0, 0.5],  # Slightly highlight Android
    wedgeprops={'edgecolor': 'white', 'linewidth': 1.5},
)

plt.title('Mobile OS Market Share (Pie Chart)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
os = ["A", "I", "O"]
ms = [75.0, 20.0, 5.0]
colors = ["#538be6", "#24a0ed", "#206b96"]

plt.pie(
    ms, 
    labels=os,
    autopct='%1.2f%%',
    colors=colors,
    explode=[0.05,0, 0.5],
    shadow=True,
    startangle= 140,  # Slightly highlight Android
    wedgeprops={'edgecolor': 'white', 'linewidth': 1.5, 'linestyle': '--'}
    
)
plt.title('Mobile OS Market Share (Pie Chart)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

```

```python

# ==========================================
# 2. Horizontal Bar Chart
# ==========================================
plt.figure(figsize=(8, 4))
bars = plt.barh(os_names, market_share, color=colors)

# Display numerical percentage values at the end of each bar
for bar in bars:
  width = bar.get_width()
  plt.text(
      width + 1,
      bar.get_y() + bar.get_height() / 2,
      f'{width}%',
      va='center',
      fontsize=10,
      fontweight='bold',
  )

plt.xlabel('Market Share (%)', fontsize=11)
plt.ylabel('Mobile OS', fontsize=11)
plt.title(
    'Mobile OS Market Share (Horizontal Bar Chart)',
    fontsize=14,
    fontweight='bold',
)
plt.xlim(0, 85)  # Extra room for value labels
plt.gca().invert_yaxis()  # Keep top category (Android) at the top
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# X-axis timeline
days = [1, 2, 3, 4, 5]

# Y-axis data components
working = [8, 8, 7, 9, 6]
sleeping = [7, 6, 8, 7, 9]
relaxing = [9, 10, 9, 8, 9]

# Create stack plot
plt.stackplot(days, working, sleeping, relaxing, labels=['Working', 'Sleeping', 'Relaxing'])

plt.title('Daily Time Distribution')
plt.xlabel('Days')
plt.ylabel('Hours')
plt.legend(loc='upper left')

plt.show()
```
