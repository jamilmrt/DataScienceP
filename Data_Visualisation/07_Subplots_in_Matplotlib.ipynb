# Subplots in Matplotlib
- Subplots allow you to show multiple plots in a single figure, side-by-side or in a grid layout. This is helpful when comparing different datasets or aspects of the same data.

```python
import matplotlib.pyplot as plt 
```

```python
x = [1,2,3,4,5]
y1 = [i*2 for i in x]
y2 = [i**2 for i in x]
```

```python
plt.subplot(1, 2, 1)
plt.plot(x, y1)
plt.title("Double of x")
plt.xlabel("Number")
plt.ylabel("Value")

plt.subplot(1, 2, 2)
plt.plot(x, y2)
plt.title("Square of x")
plt.xlabel("Number")
plt.ylabel("Value")
```

```python
# More variations of x
y3 = [i ** 0.5 for i in x]
y4 = [10 - i for i in x]

plt.figure(figsize=(8, 6))  # Optional: make it bigger

plt.subplot(2, 2, 1)
plt.plot(x, y1)
plt.title('x * 2')

plt.subplot(2, 2, 2)
plt.plot(x, y2)
plt.title('x squared')

plt.subplot(2, 2, 3)
plt.plot(x, y3)
plt.title('sqrt(x)')

plt.subplot(2, 2, 4)
plt.plot(x, y4)
plt.title('10 - x')

plt.tight_layout()
plt.show()
```

```python
fig, axs = plt.subplots(2, 2, figsize=(10, 4))
axs[0, 0].plot(x, y1)
axs[0, 0].set_title('x * 2')
axs[0, 1].plot(x, y2)
axs[0, 1].set_title('x squared')
axs[1, 0].plot(x, y3)
axs[1, 0].set_title('sqrt(x)')
axs[1, 1].plot(x, y4)
axs[1, 1].set_title('10 - x')
fig.suptitle('Subplots using plt.subplots()', fontsize=16)
fig.savefig('subplots_example.png')    # Save the figure
```

```python
fig, axs = plt.subplots(2, 2, figsize=(8, 6))
ys = [y1, y2, y3, y4]

titles = ['x * 2', 'x squared', 'sqrt(x)', '10 - x']

for i in range(2):
    for j in range(2):
        idx = i * 2 + j
        axs[i, j].plot(x, ys[idx])
        axs[i, j].set_title(titles[idx])

plt.tight_layout()
plt.show()

```
