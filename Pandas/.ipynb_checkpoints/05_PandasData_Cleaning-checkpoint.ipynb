# Data Cleaning & Preprocessing
> Real-world data is messy. Pandas gives us powerful tools to clean and transform data before analysis.

```python
import pandas as pd
```

```python
df = pd.read_csv(r"data_cleaning_sample.csv")
```

```python
df
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df.dropna()
```

```python
df.dropna(axis=1)
```

```python
df.fillna(0)
```

```python
df["Age"].fillna(df["Age"].mean())
```

```python
df
```

```python
df.fillna(method='ffill')
```

```python
df.fillna(method='bfill')
```

```python
df.duplicated()
```

```python
df.drop_duplicates()
```

```python
df.fillna(method='ffill')
```
