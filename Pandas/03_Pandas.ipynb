# Creating DataFrames
* Let’s look at different ways to create a Pandas DataFrame — the core data structure you’ll be using 90% of the time in data science

```python
import pandas as pd
```

```python
data = [["jamil", 54], ["jack", 45], ["Sheher", 55]]

data
```

```python
df = pd.DataFrame(data, columns= ["Name", "Marks"])
df
```

```python
data1 = {"a" : [4,5,6], "b": [56,34,343]}
df1 =pd.DataFrame(data1)
df1
```

```python
import numpy as np
```

```python
arr = np.array([[1,2], [3,4]])

df2 = pd.DataFrame(arr, columns=["A", "B"])
df2
```
