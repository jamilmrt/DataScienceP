```python
for i in df["Age"].apply(lambda x: re.findall('[0-9]+', str(x))):
    if len(i) < 2:
        print("Invalid age:", i)

def extract_age(age):
    age_num = re.findall('[0-9]+', str(age))
    if len(age_num)>0:
        return age_num[0]
    else: 
        return age

age = df["Age"].apply(lambda x: extract_age(x)).copy()
age[age == "nan years"] = None
```
