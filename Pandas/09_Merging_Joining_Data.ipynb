# Merging & Joining Data

```python
import pandas as pd 
```

```python
employees = pd.DataFrame({
    "EmpID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "DeptID": [10, 20, 30]
})
```

```python

departments = pd.DataFrame({
    "DeptID": [10, 20, 40],
    "DeptName": ["HR", "Engineering", "Marketing"]
})
```

```python
employees
```

```python
departments
```

```python
pd.merge(employees, departments, on="DeptID") #inner join
```

```python
pd.merge(employees, departments, how="left")
```

```python
pd.merge(employees, departments, how="right")
```

```python
pd.merge(employees, departments, how="outer")
```

```python
df1 = pd.DataFrame({"Name": ["Alice", "Bob"]})
df2 = pd.DataFrame({"Name": ["Charlie", "David"]})

```

```python
df1
```

```python
df2
```

```python
pd.concat([df1, df2], axis =1)
```

```python
df1 = pd.DataFrame({"Name": ["Alice", "Bob"], "score": [1, 2]})
df2 = pd.DataFrame({"Name": ["Charlie", "David"], "age": [45, 78]})

```

```python
pd.concat([df1, df2], axis =1)
```
