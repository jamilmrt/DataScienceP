```python
import pandas as pd
import numpy as np
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
df = pd.read_excel("iris-train.xlsx")
```

```python
df.head()
```

```python
xtrain = df.iloc[:, :-1]
ytrain = df.iloc[:,-1]
```

```python
model = RandomForestClassifier()
```

```python
model.fit(xtrain, ytrain)
```

```python
data = pd.read_excel("iris-test.xlsx")
data.head()
```

```python
xtest = data.iloc[:, :-1]
ytest = data.iloc[:, -1]
```

```python
pred = model.predict(xtest)
```

```python
pred
```

```python
# Accuracy = Correct Predictions / Total Predictions = 90 / 100 = 90%
```

```python
actual = ytest.values
```

```python
count = 0
for i in range(0, len(actual)):
    if actual[i] == pred[i]:
        count +=1
print((count * 100)/ len(actual))
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dtc = DecisionTreeClassifier()
```

```python
dtc.fit(xtrain, ytrain)
```

```python
dtcpred = dtc.predict(xtest)
```

```python
count = 0
for i in range(0, len(actual)):
    if actual[i] == dtcpred[i]:
        count +=1
print((count * 100)/ len(actual))
```
