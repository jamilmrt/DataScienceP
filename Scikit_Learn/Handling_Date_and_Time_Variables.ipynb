```python
import numpy as np
import pandas as pd
```

```python
date = pd.read_csv("https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/main/day34-handling-date-and-time/orders.csv")
time = pd.read_csv("https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/main/day34-handling-date-and-time/messages.csv")
```

```python
date.info()
```

```python
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

```python
date["date_year"] = date['date'].dt.year
date.head()
```

```python
date["date_month"] = date['date'].dt.month
date.head()
```

```python
date['date_month_name'] = date['date'].dt.month_name()
date.head()
```

```python
date['date_day'] = date['date'].dt.day
date.head()
```

```python
date['date_dow_name'] = date['date'].dt.day_name()
date.head()
```

```python
date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday', 'Saturday']), 1, 0)
date.head()
```

```python
date.drop(columns=['product_id', 'city_id', 'orders'], inplace=True)

date.head()
```

```python
# Week of the year?
date['date_weekday'] = date['date'].dt.weekday
date.head()
```

```python
time.info()
```

```python
time['date'] = pd.to_datetime(time['date'])
```

```python
time.info()
```

```python
time["hour"] = time['date'].dt.hour
time['minute'] = time['date'].dt.minute
time['second'] = time['date'].dt.second
```

```python
import datetime
today = datetime.datetime.today()
today - time['date']
```

```python
# days
(today - time['date']).dt.days
```

```python
# Months

np.round((today - time['date'])/np.timedelta64(1, 'm'),0)
```

```python
(today - time['date'])/np.timedelta64(1, 's')
```

```python
#hour

(today - time['date'])/np.timedelta64(1, 'h')
```
