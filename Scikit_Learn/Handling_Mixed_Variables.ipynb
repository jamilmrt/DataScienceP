```python
import numbers

import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("titanic.csv")
df.head()
```

```python
df['number_numerical'] = pd.to_numeric(df['number'], errors='coerce', downcast='integer')
```

```python
df["number_categorical"] = np.where(df["number_numerical"].isnull(), df["number"],np.nan)
```

```python
df.head()
```

```python
df['number'].unique()
```

```python
fig = df['number'].value_counts().plot(kind='bar')
fig.set_title("Passengers Travelling with")
```

```python
df['Cabin'].unique()
```

```python
df['cabin_numerical'] = df["Cabin"].str.extract('(d+)')
df['cabin_categorical'] = df['Cabin'].str[0]
df.head()
```

```python
df['ticket_num'] = df["Ticket"].apply(lambda s: s.split()[-1])
df["ticket_cat"] = df["Ticket"].apply(lambda s: " ".join(s.split()[:-1]))

df["ticket_cat"] = np.where(df['Ticket'].str.isdigit(), np.nan, df["ticket_cat"])
df.head(20)
```
