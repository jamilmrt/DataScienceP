```python

import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/main/day44-outlier-detection-using-percentiles/weight-height.csv")
```

```python
df.shape
```

```python
df['Height'].describe()
```

```python
sns.displot(df['Height'], kde=True)
```

```python
sns.boxenplot(df['Height'])
```

```python
upper_limit = df['Height'].quantile(0.99)
print("Upper",upper_limit)

lower_limit = df['Height'].quantile(0.01)
print("Lower",lower_limit)
```

```python
new_df = df[(df['Height'] <= 74.78) & (df['Height'] >= 58.13)]
new_df['Height'].describe()
```

```python
df[(df['Height'] <= 74.78) & (df['Height'] >= 58.13)]
```

```python
sns.distplot(new_df['Height'])
```

```python
sns.boxplot(new_df['Height'])
```

```python
# Capping --> Winsorization
df['Height'] = np.where(df['Height'] >= upper_limit,
        upper_limit,
        np.where(df['Height'] <= lower_limit,
        lower_limit,
        df['Height']))
```
