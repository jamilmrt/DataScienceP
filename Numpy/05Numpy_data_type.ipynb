# Data Types in NumPy

```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 4, 5], dtype='int32')
arr.dtype
```

```python
arr.dtype
```

```python
arr.astype('int32')
```

```python
arr.nbytes
```

```python
arr_large = np.array([1000000, 2000000, 3000000], dtype=np.int64)
arr_small = arr_large.astype(np.int32)  # Downcasting to a smaller dtype
print(arr_small)  # Output: [1000000 2000000 3000000]
print(arr_small.dtype)  # Output: int32
```

```python
arr_small.nbytes
```

```python
arr_int64 = np.array([1, 2, 3], dtype=np.int64)
arr_int32 = np.array([1, 2, 3], dtype=np.int32)

print(arr_int64.nbytes)  # Output: 24 bytes (3 elements * 8 bytes each)
print(arr_int32.nbytes)  # Output: 12 bytes (3 elements * 4 bytes each)
```

```python
arrstr = np.array(['apple', 'banana', 'cherry'], dtype='U10')  # Unicode string array
print(arrstr)
```

```python
arrstr.nbytes
```

```python
arr_com = np.array([1 + 2j, 3 + 4j, 5 + 6j], dtype='complex64')
print(arr_com)
```

```python
arr_com.nbytes
```
