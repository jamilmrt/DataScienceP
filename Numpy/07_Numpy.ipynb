# Built in Mathematical Functions in NumPy

```python
import numpy as np
```

```python
arr = np.array([1,2,3,4])
```

```python
np.mean(arr)
```

```python
np.median(arr)
```

```python
np.std(arr)
```

```python
np.var(arr)
```

```python
np.min(arr)
```

```python
np.max(arr)
```

```python
np.power(arr, 2)
```

```python
np.sum(arr)
```

```python
np.prod(arr)
```

```python
np.median(arr)
```

```python
np.percentile(arr, 50)  # For the 50th percentile (median)
```

```python
np.argmin(arr)
```

```python
np.argmax(arr)
```

```python
arr1 = np.array([1,2,3,4])
arr2 = np.array([5,6,7,9])
np.corrcoef(arr1, arr2)
```

```python
np.unique(arr)
```

```python
np.diff(arr)
```

```python
np.cumsum(arr)
```
