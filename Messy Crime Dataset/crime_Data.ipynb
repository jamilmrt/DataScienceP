```python
# STEP 2: Map the remaining variations to standard Master Categories
crime_mapping = {
    # Assault & Battery (Non-Sexual, Non-Domestic)
    'Asslt': 'Assault',
    'Battery': 'Assault',
    'Assault & Battery': 'Assault',
    
    # Sexual Assault
    'Sexual Assualt': 'Sexual Assault',
    'Sex Assault': 'Sexual Assault',
    'Sa': 'Sexual Assault',
    
    # Domestic Violence
    'Domestc Violence': 'Domestic Violence',
    'Dom. Violence': 'Domestic Violence',
    'Dv': 'Domestic Violence',
    
    # Theft & Larceny
    'Theft/Larceny': 'Theft',
    'Larceny': 'Theft',
    'Stealing': 'Theft',
    
    # Burglary & B&E
    'Burglry': 'Burglary',
    'B&E': 'Burglary',
    'Breaking & Entering': 'Burglary',
    
    # Robbery
    'Robbry': 'Robbery',
    'Roberry': 'Robbery',
    'Armed Robbery': 'Robbery',
    
    # Homicide
    'Homocide': 'Homicide',
    'Murder': 'Homicide',
    'Manslaughter': 'Homicide',
    
    # DUI / Drunk Driving
    'Drunk Driving': 'DUI',
    'Duii': 'DUI',
    'Dwi': 'DUI',
    'D.U.I.': 'DUI',
    'Dui': 'DUI',
    
    # Drugs & Narcotics
    'Drug Offense': 'Drugs/Narcotics',
    'Drug Offence': 'Drugs/Narcotics',
    'Narcotics': 'Drugs/Narcotics',
    'Drugs': 'Drugs/Narcotics',
    
    # Fraud & Deception
    'Fraudulent Activity': 'Fraud',
    'Scam': 'Fraud',
    'Deception': 'Fraud',
    'Online Fraud': 'Fraud',
    
    # Cybercrime
    'Cyber Crime': 'Cybercrime',
    'Hacking': 'Cybercrime',
    
    # Kidnapping & Abduction
    'Kidnaping': 'Kidnapping',
    'Abduction': 'Kidnapping',
    
    # Vandalism & Property Damage
    'Property Damage': 'Vandalism',
    'Graffiti': 'Vandalism',
    'Vandlism': 'Vandalism',
    
    # Arson
    'Arsen': 'Arson',
    'Fire Setting': 'Arson',
    
    # Trespassing
    'Trespass': 'Trespassing',
    'Tresspassing': 'Trespassing'
}

# Apply the mapping mapping to the column
# We use .replace() so any category already named correctly (like 'Assault') is left alone
df['crime_type'] = df['crime_type'].replace(crime_mapping)
```

```python
# 1. Define the exact list of categories you expect to have
master_categories = [
    'Assault', 'Sexual Assault', 'Domestic Violence', 'Theft', 
    'Burglary', 'Robbery', 'Homicide', 'DUI', 'Drugs/Narcotics', 
    'Fraud', 'Cybercrime', 'Kidnapping', 'Vandalism', 'Arson', 'Trespassing'
]

# 2. Find any rows where the Crime_Type is NOT in your master list
# The ~ symbol negates the .isin() condition
unmapped_rows = df[~df['crime_type'].isin(master_categories)]

# 3. Print just the unique weird values that were found
unmapped_stragglers = unmapped_rows['crime_type'].unique()

print("Unmapped categories found:")
print(unmapped_stragglers)
```
