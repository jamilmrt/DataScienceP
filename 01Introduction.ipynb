```python
import json
```

```python
def load_data(filename):
    with open(filename, 'r') as f:
        data = json.load(f)
    return data
    
```

```python
data = load_data('data.json')
```

```python
data
```

```python
# write a function to display the user and their connection
def display_user(data):
    for user in data['users']:
        print(f"ID:{user['id']} - {user['name']} is friends with {user['friends']} and like page are {user['liked_pages']}")
    print("Page Information")
    for page in data['pages']:
        print(f'{page['id']} : {page['name']}')
display_user(data)   
              
              
    
        
```
