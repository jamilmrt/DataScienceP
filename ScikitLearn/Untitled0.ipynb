```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, LabelEncoder, OrdinalEncoder
```

```python
from sklearn.model_selection import train_test_split
X = df.iloc[:, :-1]
y = df.iloc[:, -1]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
y_train.head()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
ordinal_encoding = Pipeline(steps=[
    ('ordinal_encoder', OrdinalEncoder(categories=[['Poor', 'Average', 'Good'], ['School', 'UG', 'PG']]))
])

label_encoding = Pipeline(steps=[
    ('label_encoder', LabelEncoder())
])
```

```python
full_pipeline = ColumnTransformer(transformers=[
    ('ordinal_encoding', ordinal_encoding, ['review', 'education']),
    ('label_encoding', label_encoding, ['purchased'])
])
```
